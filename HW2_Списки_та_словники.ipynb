**1. Списки - базові операції**

Створіть список температур за тиждень (у градусах Цельсія) з назвою
`temperatures` і значеннями [22, 25, 19, 23, 26, 20, 21, 22].


Виконайте (кожен пункт оцінюється в 0.5 бала):
- Виведіть перший елемент
- Виведіть останній елемент
- Додайте температуру 24
- Видаліть температуру 19
- Виведіть довжину списку
- Обчисліть суму всіх температур
- Знайдіть максимальну температуру

```python
# 1. Списки - базові операції
temperatures = [22, 25, 19, 23, 26, 20, 21, 22]
print(f"Список температур: {temperatures}")

print(f"Перший елемент: {temperatures[0]}")
print(f"Останній елемент: {temperatures[-1]}")

temperatures.append(24)          # додаємо температуру в кінець списку
print(f"Після append(24): {temperatures}")

temperatures.remove(19)          # видаляємо перше входження значення 19
print(f"Після remove(19): {temperatures}")

print(f"Довжина списку: {len(temperatures)}")
print(f"Сума всіх температур: {sum(temperatures)}")
print(f"Максимальна температура: {max(temperatures)}")
```


**2. Словники - базові операції**

Створіть словник про себе, який містить ключі `name`, `age`, `city` і значення, які відповідають вашему імені, віку і місту (можна вставити будь-які значення, не обовʼязково свої).


Виконайте (кожен пункт оцінюється в 0.5 бала):
- Виведіть значення ключа "name"
- Змініть значення ключа "age" на 21
- Додайте новий ключ "hobby" зі значенням "programming"
- Виведіть всі ключі
- Виведіть всі значення


```python
# 2. Словники - базові операції
person = {
    "name": "Олексій",
    "age": 51,
    "city": "Київ"
}
print(f"Словник: {person}")

print(f"Значення ключа 'name': {person['name']}")

person["age"] = 21                      # змінюємо значення існуючого ключа
print(f"Після зміни age: {person}")

person["hobby"] = "programming"         # додаємо новий ключ
print(f"Після додавання hobby: {person}")

print(f"Усі ключі: {list(person.keys())}")
print(f"Усі значення: {list(person.values())}")
```
