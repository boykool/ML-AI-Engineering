**1. Функція з параметром**

Створіть функцію `show_menu()`, яка приймає перелік страв, наприклад ['Піца', 'Паста', 'Салат'] і виводить меню ресторану в наступному вигляді:
```
=== МЕНЮ ===
1. Піца
2. Паста
3. Салат
=============
```

Викличте цю функцію на списку ["Суп грибний", "Котлета по-київськи", "Лимонад"].

````python
def show_menu(menu):
    print("=== МЕНЮ ===")
    for item in menu:
        print(f"{menu.index(item) + 1}. {item}")
    print("=============")

# Виклик
menu1 = ['Піца', 'Паста', 'Салат']
show_menu(menu1)
menu2 = ["Суп грибний", "Котлета по-київськи", "Лимонад"]
show_menu(menu2)
````

**2. Функція з параметром**

Створіть функцію `print_square(n)`, яка виводить квадрат числа.

Наприклад: `print_square(5)` → "Квадрат числа 5 = 25"

Викличте її з різними числами.

````python
def print_square(n):
    result = n ** 2
    print(f"Квадрат числа {n} = {result}")

# Виклики
print_square(5)
print_square(8)
print_square(12)
````

**3. Функція з поверненням значення**

Створіть функцію `multiply(a, b)`, яка повертає добуток двох чисел.

Викличте функцію на власноруч заданих значеннях та виведіть результат.


````python
def multiply(a, b):
    return a * b

# Виклик
result = multiply(6, 7)
print(f"8 * 7 = {result}")

result2 = multiply(12, 5)
print(f"10 * 5 = {result2}")
````

**4. Функція пошуку максимумального значення**

Створіть функцію `find_max(numbers)`, яка:
- Приймає список чисел
- Повертає найбільше число в списку

Протестуйте на списку `[15, 42, 8, 23, 56, 19]`.


````python
def find_max(numbers):
    maximum = numbers[0]
    for num in numbers:
        if num > maximum:
            maximum = num
    return maximum

# Тест
test_numbers = [15, 42, 8, 23, 56, 19]
max_num = find_max(test_numbers)
print(f"Найбільше число: {max_num}")

# Ще тест
prices = [120, 340, 89, 450, 210]
max_price = find_max(prices)
print(f"Найвища ціна: {max_price}")
````

**5. Функція з умовами**

Створіть функцію `check_grade(grade)`, яка:
- Приймає оцінку (число)
- Повертає "Здав" якщо grade >= 60
- Повертає "Не здав" якщо grade < 60

Протестуйте на різних оцінках.


````python
def check_grade(grade):
    if grade >= 60:
        return "Здав"
    else:
        return "Не здав"

# Тести
print(check_grade(75))   # Здав
print(check_grade(55))   # Не здав
print(check_grade(60))   # Здав
print(check_grade(45))   # Не здав

# Можна використовувати в циклі
test_grades = [85, 50, 65, 45, 90, 55]
for grade in test_grades:
    result = check_grade(grade)
    print(f"Оцінка {grade}: {result}")
````
