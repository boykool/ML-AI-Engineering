**1. Створення змінних**

Створіть змінні для особистої інформації (кожен пункт оцінюється в 0.5 бала):
- `name` - ваше ім'я (рядок)
- `age` - ваш вік (ціле число)
- `height` - ваш зріст в метрах (дробове число)
- `is_student` - чи є ви студентом (True/False)

Виведіть кожну з цих змінних за допомогою функції `print()`.



```python
# 1. Створення змінних
name = "Олексій"          # рядок (str)
age = 51                   # ціле число (int)
height = 2.05              # дробове число (float), кома в Python пишеться крапкою
is_student = True          # логічне значення (bool)

print(name)
print(age)
print(height)
print(is_student)

# Перевіримо типи змінних
print(type(name), type(age), type(height), type(is_student))
```

**2. Арифметичні операції**

Створіть дві змінні: `var1` зі значенням 15, `var2` зі значенням 4.

Обчисліть та виведіть (кожен пункт оцінюється в 0.5 бала):
- Суму
- Різницю
- Добуток
- Ділення
- Цілочисельне ділення
- Залишок від ділення
- Піднесення до степеня


```python
# 2. Арифметичні операції
var1 = 15
var2 = 4

print(f"Сума: {var1} + {var2} = {var1 + var2}")
print(f"Різниця: {var1} - {var2} = {var1 - var2}")
print(f"Добуток: {var1} * {var2} = {var1 * var2}")
print(f"Ділення: {var1} / {var2} = {var1 / var2}")
print(f"Цілочисельне ділення: {var1} // {var2} = {var1 // var2}")
print(f"Залишок від ділення: {var1} % {var2} = {var1 % var2}")
print(f"Піднесення до степеня: {var1} ** {var2} = {var1 ** var2}")
```
