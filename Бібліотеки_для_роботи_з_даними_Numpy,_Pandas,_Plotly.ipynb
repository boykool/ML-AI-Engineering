### Класи в Python

У NumPy `np.array` - це **клас**, який дозволяє працювати з багатовимірними масивами. На відміну від `list`, `np.array` містить всі елементи **одного типу**!

**Що взагалі таке класи в Python?**

Фактично це об'єкти з певним набором функціоналу (методів і атрибутів).

**Простими словами:**

Клас - це як **шаблон** або **каркас** сутності без конкретних даних. Наприклад, каркас автомобіля передбачає, що в нього є двигун, колеса, кермо, що автомобіль вміє їздити. А конкретні автомобілі (Toyota Camry, BMW X3) - це **об'єкти**, створені за цим шаблоном.

**Основні поняття:**

- **Клас** = шаблон/каркас (Student, Car, DataFrame)
- **Об'єкт (екземпляр класу)** = конкретна сутність, створена за шаблоном (Toyota Camry, BMW X3)
- **Атрибут** = характеристика/властивість об'єкта (ім'я, вік, колір)
- **Метод** = дія/функція, яку може виконати об'єкт (їхати, говорити, обчислити середнє значення)



````python
# list - це клас
my_list = [1, 2, 3]        # my_list - об'єкт класу list
my_list.append(4)          # append - МЕТОД (дія)
length = len(my_list)      # length - результат функції
````

Ми в рамках нашого прискореного інтро в Python не будемо глибоко занурюватись у класи (бо вам не скоро доведеться писати свої класи), але якщо ви хочете - можу порадити цей [ресурс](https://pynative.com/python/object-oriented-programming/) як стартову точку.

Натомість, ми будемо багато використовувати класи з зовнішніх бібліотек. І аби ви краще розуміли, як всередині побудований клас -
розглянемо приклад простого класу:



````python
# Створюємо свій клас (шаблон)
class Student:
    def __init__(self, name, age):
        self.name = name      # АТРИБУТ - характеристика студента
        self.age = age        # АТРИБУТ - характеристика студента

    def introduce(self):      # МЕТОД - дія, яку може виконати студент
        print(f"Привіт, я {self.name}, мені {self.age} років")

    def have_birthday(self):  # МЕТОД - ще одна дія
        self.age += 1
        print(f"Ура! Тепер мені {self.age}!")
````

````python
# Створюємо ОБ'ЄКТИ (конкретних студентів) за шаблоном
student1 = Student("Олег", 20)   # ОБ'ЄКТ 1
student2 = Student("Анна", 22)   # ОБ'ЄКТ 2
````

````python
student1.name, student2.name
````

````python
student2.introduce()
````

````python
# Викликаємо МЕТОДИ (виконуємо дії)

student1.have_birthday()  # метод - Ура! Тепер мені 21!
````

````python
student1.introduce()
````


**NumPy - швидкий для роботи з векторами і матрицями**

NumPy операції **в 10-100 разів швидші** за звичайні цикли Python через:

1. **Векторизація** - операція виконується одразу для всього масиву, а не по одному елементу
2. **Код на C** - NumPy написаний на C (швидкій мові), а не чистому Python
3. **Без циклів Python** - не треба перевіряти типи на кожній ітерації


````python
import numpy as np
import time

# Створюємо великий список/масив
size = 1_000_000
python_list = list(range(size))
numpy_array = np.arange(size)
````

## Візуалізація даних

### Pandas Plotting (вбудована візуалізація)

Pandas має вбудовану візуалізацію, яка працює **на базі Matplotlib** (головної бібліотеки для графіків в Python). Оскільки в Data Science ми багато працюємо з Pandas DataFrame, це **найпростіший спосіб** швидко побудувати графіки - просто викликаємо `.plot()` на наших даних!

**Переваги:**
- ✅ Не треба конвертувати дані в інший формат - працюємо безпосередньо з DataFrame
- ✅ Один рядок коду замість пʼяти
- ✅ Ідеально для швидкого аналізу даних

**Важливо:** Pandas plotting використовує Matplotlib під капотом, тому всі налаштування Matplotlib (стилі, розміри, кольори) працюють і тут!


````python
import matplotlib.pyplot as plt
````

Створювати візуалізації з Pandas - супер просто.
1. Розуміємо, яку колонку хочемо візуалізувати.
2. Розуміємо, з яким графіком хочемо відобразити дані.
3. Викликаємо правильний метод згідно [документації](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.plot.html).

Графіки, доступні для побудови з Pandas Plotting (взято з документації):

**kind** (str) - Тип графіка:

* `'line'` : лінійний графік (за замовчуванням)
* `'bar'` : вертикальна стовпчикова діаграма
* `'barh'` : горизонтальна стовпчикова діаграма
* `'hist'` : гістограма (розподіл)
* `'box'` : ящик з вусами (boxplot) - показує медіану, квартилі, викиди
* `'kde'` : графік щільності ядра (Kernel Density Estimation)
* `'density'` : те ж саме, що `'kde'`
* `'area'` : діаграма з областю
* `'pie'` : кругова діаграма
* `'scatter'` : точковий графік (тільки для DataFrame)
* `'hexbin'` : шестикутна діаграма щільності (тільки для DataFrame)

Якщо не знайомі з якимось типом графіку - в цьому сервісі є хороші описи кожного: https://www.data-to-viz.com/

Побудуємо кілька простих графіків, а потім додамо більше оформлення.


### Plotly - Інтерактивні графіки
Іноді зручніше дослідити дані на інтерактивному графіку. Особливо, коли багато точок на графіку (багато стовпців, або точок на scatter plot).

Для побудови інтерактивних графіків в Python використовують найчастіше бібліотеку Plotly. В неї є підмодуль express для швидшого створення базових візуалізацій.

Документація з переліком графіків, які доступні в Plotly: https://plotly.com/python/

````python
import plotly.express as px
import plotly.graph_objects as go
````

````python
# Інтерактивна гістограма
fig = px.histogram(df, x='age', nbins=30,
                   title='Розподіл віку пасажирів (інтерактивно)',
                   labels={'age': 'Вік'})
fig.show()
````

````python
# Інтерактивний scatter plot
fig = px.scatter(df, x='age', y='fare', color='survived',
                #  size='family_size',
                 hover_data=['name', 'pclass'],
                 title='Вік vs Вартість квитка',
                 labels={'survived': 'Вижив', 'age': 'Вік', 'fare': 'Вартість'},
                 color_continuous_scale='RdYlGn')
fig.show()
````

````python
fig = px.box(df, x='survived', y='family_size')
fig.show()
````

````python
# Box plot з Plotly
fig = px.box(df, x='pclass', y='age', color='sex',
             title='Розподіл віку за класом та статтю',
             labels={'pclass': 'Клас', 'age': 'Вік'})
fig.show()
````

````python
# Sunburst - ієрархічна візуалізація
df_hierarchy = df.groupby(['pclass', 'sex', 'survived']).size().reset_index(name='count')
fig = px.sunburst(df_hierarchy,
                  path=['pclass', 'sex', 'survived'],
                  values='count',
                  title='Ієрархія виживання: Клас → Стать → Результат')
fig.show()
````

````python
import numpy as np
````

````python
# Heatmap кореляції
numeric = df.select_dtypes(include=[np.number])
correlation = numeric.corr()

fig = px.imshow(correlation,
                text_auto='.2f',
                aspect='auto',
                title='Кореляційна матриця',
                color_continuous_scale='RdBu_r')
fig.show()
````

````python
# 3D scatter plot
fig = px.scatter_3d(df,
                    x='age', y='fare', z='pclass',
                    color='survived',
                    hover_data=['name'],
                    title='3D візуалізація: Вік, Вартість, Клас',
                    labels={'survived': 'Вижив'})
fig.show()
````
