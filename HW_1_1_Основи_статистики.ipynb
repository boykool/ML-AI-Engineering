**Завдання 2.** Ми зібрали дані про суму покупок в гривнях 20ма випадково обраними клієнтами в нашому магазині (всього клієнтів у нас більше) за минулий місяць. Вони представлені в змінній `sum_purchases`. Знайдіть **вибіркове середнє** для цих даних кожним з зазначених методів:   

  2.1. без використання жодних бібліотек, користуючись тільки Python.  
  2.2. використовуючи відповідний метод бібліотеки `numpy`.

  Виведіть результат обчислень в кожному з випадків на екран та порівняйте результати. Вони співпадають?

```python
sum_purchases = [ 316,  493,  367,  153,  224,  121,  322,  450,  196,  397,  281,
  239,  484,  124,  223,  448,  432,  255, 4310, 3524]
```

```python
import numpy as np

# 2.1. Чистий Python
mean_purchases = sum(sum_purchases) / len(sum_purchases)
print('Середнє (Python):', mean_purchases)

# 2.2. numpy
print('Середнє (numpy):', np.mean(sum_purchases))

print('Результати співпадають:', mean_purchases == np.mean(sum_purchases))
```

**Завдання 3.** Для того ж набору даних в змінній `sum_purchases` обчисліть **медіану** кожним з зазначених методів:  

  3.1. без використання жодних бібліотек, користуючись тільки Python.  
  3.2. використовуючи відповідний метод бібліотеки numpy.  

Виведіть результат обчислень в кожному з випадків на екран. Порівняйте результати зі значенням вибіркового середнього. Напишіть коментар, для яких задач ви б використовували вибіркове середнє, а для яких - медіану?

```python
# 3.1. Чистий Python
sorted_purchases = sorted(sum_purchases)
n = len(sorted_purchases)
if n % 2 == 0:
    median_purchases = (sorted_purchases[n//2 - 1] + sorted_purchases[n//2]) / 2
else:
    median_purchases = sorted_purchases[n//2]
print('Медіана (Python):', median_purchases)

# 3.2. numpy
print('Медіана (numpy):', np.median(sum_purchases))

print('Середнє:', np.mean(sum_purchases))
```

**Висновок:** медіана (319 грн) суттєво менша за середнє (667.95 грн). Причина - два викиди (4310 і 3524 грн), які «тягнуть» середнє вгору, а на медіану майже не впливають.

- **Вибіркове середнє** доцільно використовувати, коли дані розподілені приблизно симетрично і без викидів (наприклад, зріст людей, температура, оцінки за тест), а також коли важлива загальна сума (середнє × кількість = сума, напр. для планування виторгу).
- **Медіану** доцільно використовувати, коли в даних є викиди або розподіл скошений (доходи, ціни на нерухомість, суми покупок, час відповіді сервера) - вона краще описує «типове» значення.

**Завдання 4**. Обчислість стандартне відхилення наведеної вибірки в змінній `sum_purchases` кожним з зазначених методів:  

  4.1. без використання жодних бібліотек, користуючись тільки Python.  
  4.2. використовуючи відповідний метод бібліотеки numpy.  

Виведіть результат обчислень в кожному з випадків на екран.

```python
# 4.1. Чистий Python (вибіркове стандартне відхилення, ділимо на n-1)
purchases_minus_mean = [(el - mean_purchases)**2 for el in sum_purchases]
std_purchases_sample = (sum(purchases_minus_mean) / (len(sum_purchases) - 1)) ** 0.5
print('Стандартне відхилення (Python):', std_purchases_sample)

# 4.2. numpy
print('Стандартне відхилення (numpy):', np.std(sum_purchases, ddof=1))
```

**Завдання 5**. Ми провели дослідження віку студентів-першокурсників. Виявилось, що
- 10 студентів мають вік 16 років
- 25 студентів - 17 років
- 20 студентів - 18 років
- 2 студенти - 20 років

Обчислість середнє значення для цієї вибірки будь-яким способом - на чистому Python або з numpy - та виведіть результат на екран.

```python
ages = [16, 17, 18, 20]
counts = [10, 25, 20, 2]

mean_age = sum([ages[i] * counts[i] for i in range(len(ages))]) / sum(counts)
print('Середній вік (Python):', mean_age)

print('Середній вік (numpy):', np.average(ages, weights=counts))
```

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
# 6.1. Чистий Python
mean_visits = sum(visits) / len(visits)

visits_minus_mean = [(el - mean_visits)**2 for el in visits]
std_visits_sample = (sum(visits_minus_mean) / (len(visits) - 1)) ** 0.5

xy = [(sum_purchases[i] - mean_purchases) * (visits[i] - mean_visits) for i in range(len(visits))]
cov_xy = sum(xy) / (len(visits) - 1)

corr_python = cov_xy / (std_purchases_sample * std_visits_sample)
print('Коефіцієнт кореляції Пірсона (Python):', corr_python)

# 6.2. numpy
print('Коефіцієнт кореляції Пірсона (numpy):', np.corrcoef(sum_purchases, visits)[0, 1])
```

```python
import matplotlib.pyplot as plt

plt.scatter(sum_purchases, visits)
plt.title('Scatter Plot of Purchases vs. Visits')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```

**Аналіз:** коефіцієнт кореляції ≈ -0.30 - слабкий від'ємний лінійний зв'язок між сумою покупок і кількістю відвідувань.

На графіку розсіювання видно, що два клієнти з дуже великими сумами покупок (4310 і 3524 грн) - це потенційні викиди, які далеко відстоять від решти точок. Коефіцієнт Пірсона дуже чутливий до таких значень: дві точки можуть суттєво змінити його величину. Щоб оцінити вплив цих двох точок, порівняємо кореляцію для повної вибірки та після їх вилучення.

Що варто змінити: спершу дослідити ці записи (помилка введення чи реальні великі покупки, напр. оптовий клієнт). Якщо це помилки - виправити або вилучити. Якщо це реальні покупки - сам факт великої суми не робить їх «неправильними», а їх вилучення змінює досліджувану групу клієнтів (кореляція тоді описує лише «звичайних» покупців). Тому корисно порахувати кореляцію з цими записами і без них та порівняти результати, або використати стійкіший до викидів коефіцієнт Спірмена.

**Завдання 7**. В попередньому завданні ми помітили, що викиди нам заважають отримати "правдиве" значення кореляції. Давайте вилучимо записи, які відповідають значенням з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте `threshold`, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).

Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою `numpy` для кількості візитів і суми покупок без записів-викидів. Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

```python
threshold = np.mean(sum_purchases) + 2 * np.std(sum_purchases, ddof=1)
print('threshold:', threshold)

purchases_clean = [sum_purchases[i] for i in range(len(sum_purchases)) if sum_purchases[i] <= threshold]
visits_clean = [visits[i] for i in range(len(sum_purchases)) if sum_purchases[i] <= threshold]
print('Кількість записів без викидів:', len(purchases_clean))

print('Коефіцієнт кореляції Пірсона без викидів:', np.corrcoef(purchases_clean, visits_clean)[0, 1])

plt.scatter(purchases_clean, visits_clean)
plt.title('Scatter Plot of Purchases vs. Visits (without outliers)')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```

**Аналіз:** за умовою завдання вилучили 2 записи з сумою покупок вище threshold ≈ 2917 грн. Порівняння: для всіх 20 клієнтів r ≈ -0.303, для решти 18 клієнтів r ≈ -0.369 - слабкий/помірний від'ємний лінійний зв'язок. Це не «виправлення» попереднього значення, а оцінка зв'язку для іншої групи - клієнтів без надвеликих покупок.

Проте зв'язок слабкий - точки на діаграмі сильно розкидані, тож кількість відвідувань погано пояснює суму покупок, і вибірка з 18 клієнтів замала для впевнених висновків.
